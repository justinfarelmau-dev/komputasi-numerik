```

# ==================================================
# PROGRAM SPL 3 VARIABEL METODE ELIMINASI GAUSS
# ==================================================

print("========================================")
print(" PENYELESAIAN SPL 3 VARIABEL")
print("      METODE ELIMINASI GAUSS")
print("========================================")

# 1. DATA PERSAMAAN (OTOMATIS)
# 2x + y + z = 10
# x + 2y + z = 12
# x + y + 2z = 14

A = [
    [2.0, 1.0, 1.0, 10.0],
    [1.0, 2.0, 1.0, 12.0],
    [1.0, 1.0, 2.0, 14.0]
]

# 2. MENAMPILKAN PERSAMAAN
print("\nPersamaan yang digunakan:")
print("2X + Y + Z = 10")
print("X + 2Y + Z = 12")
print("X + Y + 2Z = 14")

# 3. MENAMPILKAN MATRIKS AWAL
print("\nMatriks Augmented Awal:")
for baris in A:
    print(baris)

# 4. PROSES ELIMINASI GAUSS
for i in range(3):

    # Pertukaran baris jika pivot nol
    if abs(A[i][i]) < 1e-10:
        for k in range(i + 1, 3):
            if abs(A[k][i]) > 1e-10:
                A[i], A[k] = A[k], A[i]
                break

    if abs(A[i][i]) < 1e-10:
        raise ValueError("Sistem tidak memiliki solusi tunggal")

    # Normalisasi pivot
    pivot = A[i][i]

    for j in range(4):
        A[i][j] /= pivot

    # Eliminasi elemen di bawah pivot
    for k in range(i + 1, 3):
        faktor = A[k][i]

        for j in range(4):
            A[k][j] -= faktor * A[i][j]

    print(f"\nMatriks setelah eliminasi ke-{i+1}:")
    for baris in A:
        print([round(x, 4) for x in baris])

# 5. SUBSTITUSI MUNDUR
Z = A[2][3] / A[2][2]

Y = (A[1][3] - A[1][2] * Z) / A[1][1]

X = (A[0][3] - A[0][1] * Y - A[0][2] * Z) / A[0][0]

# 6. MENAMPILKAN HASIL
print("\n========================================")
print("             HASIL AKHIR")
print("========================================")

print(f"Nilai X = {X:.2f}")
print(f"Nilai Y = {Y:.2f}")
print(f"Nilai Z = {Z:.2f}")

# 7. KESIMPULAN
print("\nKesimpulan:")
print("X = 1")
print("Y = 3")
print("Z = 5")

print("\nProgram berhasil dijalankan!")
print("========================================")
```
